```python
spark.sql("USE CATALOG ecommerce_project")
spark.sql("USE SCHEMA sales")
```

```python
%sql
SELECT year, ROUND(SUM(total_profit), 2) AS profit
FROM agg_profit
GROUP BY year
ORDER BY year
```

```python
%sql
SELECT year,
       COALESCE(category, 'Unknown') AS category,
       ROUND(SUM(total_profit), 2) AS profit
FROM agg_profit
GROUP BY year, COALESCE(category, 'Unknown')
ORDER BY year, category
```

```python
%sql
SELECT customer_id,
       COALESCE(customer_name, 'Unknown') AS customer_name,
       ROUND(SUM(total_profit), 2) AS profit
FROM agg_profit
GROUP BY customer_id, COALESCE(customer_name, 'Unknown')
ORDER BY profit DESC
```

```python
%sql
SELECT customer_id,
       COALESCE(customer_name, 'Unknown') AS customer_name,
       year,
       ROUND(SUM(total_profit), 2) AS profit
FROM agg_profit
GROUP BY customer_id, COALESCE(customer_name, 'Unknown'), year
ORDER BY customer_name, year
```
